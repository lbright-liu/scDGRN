### 3. Identification of key genes

```python
import numpy as np
import pandas as pd
from collections import Counter
import math

df1 = pd.read_csv("./cytoscape/hEry_prediction1_3000.csv")
df2 = pd.read_csv("./cytoscape/hEry_prediction2_3000.csv")
df3 = pd.read_csv("./cytoscape/hEry_prediction3_3000.csv")

result1 = [(row[0], row[1]) for row in df1.values]
result2 = [(row[0], row[1]) for row in df2.values]
result3 = [(row[0], row[1]) for row in df3.values]

network_list = [result1, result2, result3]

g1 = list(set(list(df1["TF"].values) + list(df1["Target"].values)))
g2 = list(set(list(df2["TF"].values) + list(df2["Target"].values)))
g3 = list(set(list(df3["TF"].values) + list(df3["Target"].values)))

all_genes = list(set(g1 + g2 + g3))
print("All genes list:")
print(all_genes)
gene_len = len(all_genes)
print(f"Total number of unique genes: {gene_len}")

def calculate_entropy(network):
    all_nodes = [node for edge in network for node in edge]
    node_degrees = dict(Counter(all_nodes))
    total_nodes = len(node_degrees)
    degree_counts = dict(Counter(node_degrees.values()))
    entropy = 0
    for degree in degree_counts.values():
        probability = degree / total_nodes
        entropy -= probability * math.log2(probability)
    return entropy

def remove_gene_and_edges(network, gene):
    modified_network = [edge for edge in network if gene not in edge]
    return modified_network

entropy_list = [calculate_entropy(network) for network in network_list]
mean_entropy = np.mean(entropy_list)
std_entropy = np.std(entropy_list)

gene_scores = {}
for gene in all_genes:
    #print("Processing gene...")
    perturbation_values = []
    for i, network in enumerate(network_list):
        modified_network = remove_gene_and_edges(network, gene)
        modified_entropy = calculate_entropy(modified_network)
        perturbation_values.append(modified_entropy)
    
    mean_perturbation = np.mean(perturbation_values)
    std_perturbation = np.std(perturbation_values)
    R = abs(mean_perturbation - mean_entropy) * abs(std_perturbation - std_entropy)
    gene_scores[gene] = R
    

sorted_genes = sorted(gene_scores.items(), key=lambda x: x[1], reverse=True)
print("Top 20 genes by perturbation score:")
print(sorted_genes[:20])

df = pd.DataFrame(sorted_genes, columns=['gene','R_score'])
print("Full gene perturbation scores dataframe:")
print(df)

gene_index = df['gene'][-100:-50].values.tolist()
print("Gene index (rows -100 to -50):")
print(gene_index)

df_gene = pd.read_csv("./Target.csv")
dic_index_gene = df_gene.set_index("index")["Gene"].to_dict()
dic_gene_index = df_gene.set_index("Gene")["index"].to_dict()

print("Gene index (repeated):")
print(gene_index)

print("Full dataframe (repeated):")
print(df)

df_top = df.head(100)
print("Top 100 genes by perturbation score:")
print(df_top)

## draw figure
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.cm as cm
from matplotlib.colors import Normalize
from matplotlib import rcParams
import numpy as np
df_tf = pd.read_csv("TF.csv")
df_target = pd.read_csv("Target.csv")
dic_index_gene = df_target.set_index("index")["Gene"].to_dict()

#sorted_df = pd.read_csv("./saved_data/key_gene_entropy.csv")
sorted_df = df_top

gene_list = []
for i in list(sorted_df["gene"].values):
    gene_list.append(i)
sorted_df["gene"] = gene_list

top_20_genes = sorted_df['gene'].head(20)
top_20_genes = top_20_genes.str.upper()
#top_20_genes.to_csv("./saved_data/top20_gene_pertur.csv",index=False)

top_20_scores = sorted_df['R_score'].head(20)
top_20_scores = np.log(top_20_scores * 100000)

fig, ax = plt.subplots()
bars = ax.barh(top_20_genes, top_20_scores, color='#b295c5')

print("Top 20 genes:")
print(top_20_genes)
print("Top 20 genes list:")
print(list(top_20_genes))
print("Top 20 perturbation scores (log scaled):")
print(top_20_scores)

ax.invert_yaxis()
ax.set_xlabel('Perturbation Score', fontsize=12)
ax.margins(y=0.01)
ax.tick_params(axis='y', labelsize=12)
ax.tick_params(axis='x', labelsize=12)
ax.set_title('CellEvo', fontsize=15)

fig.set_size_inches(4, 6)
plt.tight_layout()
plt.show()
```

### 4. Visualization of gene expression distribution

```python
import pandas as pd
import numpy as np
import umap
import matplotlib.pyplot as plt
from matplotlib import rcParams
from sklearn.preprocessing import StandardScaler

data = pd.read_csv("log_normalization_exp_data_target.csv", index_col=0)

scaler = StandardScaler()
data = pd.DataFrame(scaler.fit_transform(data), columns=data.columns, index=data.index)

print("Normalized gene expression data:")
print(data)

df_early = pd.read_csv("Ery_expression0.csv",index_col=0)
df_mid = pd.read_csv("Ery_expression1.csv",index_col=0)
df_late = pd.read_csv("Ery_expression2.csv",index_col=0)

cell_types = {
    "Early": list(df_early.columns),
    "Mid": list(df_mid.columns),
    "Late": list(df_late.columns),
}

labels = []
for col in data.columns:
    for cell_type, cells in cell_types.items():
        if col in cells:
            labels.append(cell_type)
            break
    else:
        labels.append("Unknown")

assert len(labels) == data.shape[1]

data_t = data.T

reducer = umap.UMAP(n_neighbors=15, min_dist=0.1, random_state=66)
embedding = reducer.fit_transform(data_t)

unique_labels = list(set(labels))
colors = plt.cm.tab10(np.linspace(0, 1, len(unique_labels)))
color_map = dict(zip(unique_labels, colors))

plt.figure(figsize=(7, 6))
for label in unique_labels:
    idx = [i for i, l in enumerate(labels) if l == label]
    plt.scatter(
        embedding[idx, 0],
        embedding[idx, 1],
        label=label,
        color=color_map[label],
        s=10,
        alpha=0.8,
    )

plt.legend(loc="best", markerscale=1.5, fontsize=15)
plt.title("UMAP of erythroid", fontsize=20)
plt.xlabel("UMAP1", fontsize=20)
plt.ylabel("UMAP2", fontsize=20)
plt.xticks(fontsize=13)
plt.yticks(fontsize=13)
plt.tight_layout()
plt.show()



## show sepecific TF 
import pandas as pd
import numpy as np
import umap
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap
from matplotlib import rcParams
from sklearn.preprocessing import StandardScaler
from matplotlib import cm
from matplotlib.colors import Normalize


data = pd.read_csv("log_normalization_exp_data_target.csv", index_col=0)

scaler = StandardScaler()
data = pd.DataFrame(scaler.fit_transform(data), columns=data.columns, index=data.index)

gene_of_interest = "JUND"
print("Normalized gene expression matrix:")
print(data)

data_t = data.T

reducer = umap.UMAP(n_neighbors=15, min_dist=0.1, random_state=66)
embedding = reducer.fit_transform(data_t)

expression = data.loc[gene_of_interest].values

purples_cmap = cm.get_cmap('Purples', 256)
newcolors = purples_cmap(np.linspace(0.2, 1.0, 256))
newcolors[0] = [1, 1, 1, 1]
custom_purples = LinearSegmentedColormap.from_list('custom_purples', newcolors)

plt.figure(figsize=(8, 6))
sc = plt.scatter(
    embedding[:, 0],
    embedding[:, 1],
    c=expression,
    cmap=custom_purples,
    s=10,
    alpha=0.8,
)
plt.colorbar(sc, shrink=0.4, aspect=6)
plt.title(f"{gene_of_interest} expression", fontsize=20)
plt.xlabel("UMAP1", fontsize=20)
plt.ylabel("UMAP2", fontsize=20)
plt.xticks(fontsize=13)
plt.yticks(fontsize=13)
plt.tight_layout()
plt.show()
```

### 6. The discovery of the co-regulator

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.spatial.distance import squareform
from scipy.cluster.hierarchy import linkage, leaves_list
import numpy as np
import time
from matplotlib import rcParams
from sklearn.preprocessing import MultiLabelBinarizer


def calculate_tf_overlap_heatmap(csv_path, figsize=(20, 18), title="TF regulatory network overlap ratio heatmap",
                                 cmap='RdBu_r', save_path=None, min_targets=2):
    print(f"Reading data: {csv_path}")
    try:
        df = pd.read_csv(csv_path)
        if 'TF' not in df.columns or 'Target' not in df.columns:
            print("'TF' and 'Target' columns not found, trying to read without header...")
            df = pd.read_csv(csv_path, header=None, names=['TF', 'Target'])
    except Exception as e:
        print(f"Error reading file: {e}")
        return None, None, None, None

    print("Building TF-target gene mapping...")
    tf_target_dict = df.groupby('TF')['Target'].apply(set).to_dict()

    filtered_tf_target_dict = {tf: targets for tf, targets in tf_target_dict.items()
                               if len(targets) >= min_targets}

    filtered_out_tfs = set(tf_target_dict.keys()) - set(filtered_tf_target_dict.keys())
    print(f"Filtered out {len(filtered_out_tfs)} TFs with fewer than {min_targets} target genes")
    if filtered_out_tfs:
        print(f"Filtered TFs: {', '.join(filtered_out_tfs)}")

    tfs = list(filtered_tf_target_dict.keys())
    num_tfs = len(tfs)

    if num_tfs == 0:
        print("No qualified TF data found after filtering")
        return None, None, None, None

    print(f"Retained {num_tfs} qualified TFs")

    print("Calculating overlap ratio matrix...")
    mlb = MultiLabelBinarizer()
    binary_matrix = mlb.fit_transform([list(targets) for targets in filtered_tf_target_dict.values()])

    intersection_matrix = binary_matrix.dot(binary_matrix.T)
    target_counts = np.sum(binary_matrix, axis=1)
    union_matrix = target_counts.reshape(-1, 1) + target_counts.reshape(1, -1) - intersection_matrix

    overlap_matrix = np.divide(
        intersection_matrix,
        union_matrix,
        out=np.zeros_like(intersection_matrix, dtype=float),
        where=union_matrix != 0
    )

    print("Performing hierarchical clustering...")
    distance_matrix = 1 - overlap_matrix
    condensed_distance = squareform(distance_matrix)
    linkage_result = linkage(condensed_distance, method='average')
    clustered_indices = leaves_list(linkage_result)

    overlap_matrix_clustered = overlap_matrix[clustered_indices][:, clustered_indices]
    tfs_clustered = [tfs[i] for i in clustered_indices]
    print(overlap_matrix_clustered)

    print("Plotting heatmap...")
    plt.figure(figsize=figsize)

    tick_font_size = max(6, 15 - num_tfs // 20)

    hm = sns.heatmap(
        overlap_matrix_clustered,
        annot=False,
        fmt=".2f",
        xticklabels=tfs_clustered,
        yticklabels=tfs_clustered,
        cmap=cmap,
        vmin=0,
        vmax=1,
        linewidths=0.5,
        cbar_kws={'label': 'Overlap ratio (intersection of target genes / union of target genes)'}
    )

    cbar = hm.collections[0].colorbar
    cbar.set_label('Overlap ratio (intersection of target genes / union of target genes)', fontsize=20)
    cbar.ax.tick_params(labelsize=20)

    plt.title(title, fontsize=25)
    plt.xlabel('t3', fontsize=25)

    plt.xticks(rotation=90, ha='center', fontsize=tick_font_size)
    plt.yticks(fontsize=tick_font_size)

    ax = plt.gca()
    ax.set_xticks(np.arange(len(tfs_clustered)) + 0.5)
    ax.set_yticks(np.arange(len(tfs_clustered)) + 0.5)

    ax.set_xticklabels(tfs_clustered)
    ax.set_yticklabels(tfs_clustered)

    plt.tight_layout()



    return plt, overlap_matrix_clustered, tfs_clustered, filtered_out_tfs

if __name__ == "__main__":
    csv_path = './cytoscape/hEry_prediction3_3000.csv'
    save_path = './regulon_cluster_t3_scDGRN.pdf'

    plt, matrix, labels, filtered_tfs = calculate_tf_overlap_heatmap(
        csv_path,
        figsize=(15, 13),
        title="scDGRN",
        cmap='RdBu_r',
        save_path=save_path,
        min_targets=5
    )

    if plt:
        plt.show()
```

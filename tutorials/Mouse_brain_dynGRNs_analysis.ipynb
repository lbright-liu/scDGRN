## Overview
This notebook provides instructions on how to use scDGRN to reconstruct dynamic GRNs. Taking the aging process of mouse brain cells (aged_opc) as an example, given the limited prior knowledge of the regulation of brain cells, the transfer learning method was adopted to infer the dynamic GRNs of mouse brain cells. First, a unified representation model was trained using the publicly available prior network from NicheNet (Browaeys et al., Nat.Methods, 2020). Subsequently, the model was fine-tuned by using the limited prior regulatory data collected about normal mouse brain cells. This method helps to reconstruct the fine-grained developmental dynamics of the cell lineage during the aging stage. The following are the input files that need to be prepared, and we have provided basic sample data in the corresponding folder:

* **./Transfer_learning/aged_opc/TF.csv**：the focused transcription factors.
* **./Transfer_learning/aged_opc/Target.csv**：target gene list, generally, the top highly variable genes are selected for analysis.
* **./Transfer_learning/aged_opc/Train_set_TL.csv**：the collected prior regulations that can be used for pre-training. The gene names need to be mapped to the corresponding index numbers of the target gene list.
* **./Transfer_learning/aged_opc/fine_tune.csv**：The collected prior regulations that can be used for fine-tuning.
* **./Transfer_learning/aged_opc/ag_opc4_expression0/1/2/3.csv**：time-series single-cell expression data. This dataset is divided into four fine-grained stages based on the literature of the dataset or through pseudo-time-series clustering.

### 3. Calculation of gene scores based on dynamic network entropy

```python
import numpy as np
import pandas as pd
from collections import Counter
import math

df2 = pd.read_csv("./mouse_brain/cytoscape/aged_opc_UMI/ag_opc_prediction1_3000_nolog2.csv")
df3 = pd.read_csv("./mouse_brain/cytoscape/aged_opc_UMI/ag_opc_prediction2_3000_nolog2.csv")
df4 = pd.read_csv("./mouse_brain/cytoscape/aged_opc_UMI/ag_opc_prediction3_3000_nolog2.csv")
df5 = pd.read_csv("./mouse_brain/cytoscape/aged_opc_UMI/ag_opc_prediction4_3000_nolog2.csv")

result2 = [(row[0], row[1]) for row in df2.values]
result3 = [(row[0], row[1]) for row in df3.values]
result4 = [(row[0], row[1]) for row in df4.values]
result5 = [(row[0], row[1]) for row in df5.values]

network_list = [result2, result3, result4, result5]

g2 = list(set(list(df2["TF"].values) + list(df2["Target"].values)))
g3 = list(set(list(df3["TF"].values) + list(df3["Target"].values)))
g4 = list(set(list(df4["TF"].values) + list(df4["Target"].values)))
g5 = list(set(list(df5["TF"].values) + list(df5["Target"].values)))

all_genes = list(set(g2 + g3 + g4 + g5))
gene_len = len(all_genes)
print(len(all_genes))

def calculate_entropy(network):
    all_nodes = [node for edge in network for node in edge]
    node_degrees = dict(Counter(all_nodes))
    total_nodes = len(node_degrees)
    degree_counts = dict(Counter(node_degrees.values()))
    entropy = 0
    for degree in degree_counts.values():
        probability = degree / total_nodes
        entropy -= probability * math.log2(probability)
    entropy = entropy / math.log2(total_nodes)
    return entropy

def remove_gene_and_edges(network, gene):
    modified_network = [edge for edge in network if gene not in edge]
    return modified_network

entropy_list = [calculate_entropy(network) for network in network_list]
print(entropy_list)

mean_entropy = np.mean(entropy_list)
std_entropy = np.std(entropy_list)

gene_scores = {}
for gene in all_genes:
    perturbation_values = []
    for i, network in enumerate(network_list):
        modified_network = remove_gene_and_edges(network, gene)
        modified_entropy = calculate_entropy(modified_network)
        perturbation_values.append(modified_entropy)

    mean_perturbation = np.mean(perturbation_values)
    std_perturbation = np.std(perturbation_values)

    R = abs(mean_perturbation - mean_entropy) * abs(std_perturbation - std_entropy)
    gene_scores[gene] = R

sorted_genes = sorted(gene_scores.items(), key=lambda x: x[1], reverse=True)
print(sorted_genes[:20])
df = pd.DataFrame(sorted_genes, columns=['gene', 'R_score'])
print(df)

gene_index = df['gene'][:20].values.tolist()
print(gene_index)
df_gene = pd.read_csv("./mouse_brain/Target.csv")
df.to_csv("./mouse_brain/cytoscape/aged_opc_UMI/key_gene_entropy_aged_opc_nolog2.csv", index=False)
```

### 5. Regulon overlap across timepoints

```python
import pandas as pd
import numpy as np
from collections import defaultdict
import matplotlib.pyplot as plt

file_paths = [
    "./mouse_brain/cytoscape/aged_opc_UMI/ag_opc_prediction1_3000_nolog2.csv",
    "./mouse_brain/cytoscape/aged_opc_UMI/ag_opc_prediction2_3000_nolog2.csv",
    "./mouse_brain/cytoscape/aged_opc_UMI/ag_opc_prediction3_3000_nolog2.csv",
    "./mouse_brain/cytoscape/aged_opc_UMI/ag_opc_prediction4_3000_nolog2.csv"
]

dfs = []
time_points = ["t1", "t2", "t3", "t4"]

for i, file_path in enumerate(file_paths):
    df = pd.read_csv(file_path)
    if len(df.columns) >= 2:
        df = df.iloc[:, :2]
        df.columns = ["TF", "Target"]
        print(f"Data for time point {time_points[i]} loaded, containing {len(df)} records")
        dfs.append(df)
    else:
        print(f"Warning: File {file_path} has insufficient columns, skipped")

all_tfs = set()
for df in dfs:
    all_tfs.update(df["TF"].unique())
all_tfs = sorted(all_tfs)
print(f"Total {len(all_tfs)} unique transcription factors found")

tf_targets = defaultdict(lambda: [set() for _ in range(len(dfs))])
for t_idx, df in enumerate(dfs):
    for _, row in df.iterrows():
        tf, target = row["TF"], row["Target"]
        tf_targets[tf][t_idx].add(target)

overlap_rates = {}
for tf in all_tfs:
    rates = []
    for i in range(len(dfs) - 1):
        t1_targets = tf_targets[tf][i]
        t2_targets = tf_targets[tf][i + 1]

        if not t1_targets and not t2_targets:
            overlap_rate = 0
        else:
            intersection = len(t1_targets.intersection(t2_targets))
            union = len(t1_targets.union(t2_targets))
            overlap_rate = intersection / union if union > 0 else 0

        rates.append(overlap_rate)

    overlap_rates[tf] = rates

threshold = 0.05
filtered_tfs = {tf: rates for tf, rates in overlap_rates.items()
                if any(rate > threshold for rate in rates)}
print(f"{len(filtered_tfs)} TFs retained after filtering (overlap rate > {threshold})")

sorted_tfs = sorted(filtered_tfs.items(),
                    key=lambda x: x[1][2],
                    reverse=True)

max_tfs = 40
if len(sorted_tfs) > max_tfs:
    sorted_tfs = sorted_tfs[:max_tfs]
    print(f"Only the top {max_tfs} TFs are displayed due to excessive quantity")

data = {tf: [rate * 1 for rate in rates] for tf, rates in sorted_tfs}

print("\nProcessed data example (ratio):")
for i, (tf, rates) in enumerate(list(data.items())[:10]):
    print(f"{tf}: {rates}")

tfs = list(data.keys())
t1_data = [val[0] for val in data.values()]
t2_data = [val[1] for val in data.values()]
t3_data = [val[2] for val in data.values()]

x = np.arange(len(tfs))
width = 0.65

color_t1 = '#7cbfa6'
color_t2 = '#909fc6'
color_t3 = '#afd568'

fig, ax = plt.subplots(figsize=(20, 8))
ax.bar(x, t3_data, width, label='t3-t4', color=color_t3)
ax.bar(x, t2_data, width, bottom=t3_data, label='t2-t3', color=color_t2)
ax.bar(x, t1_data, width, bottom=np.array(t2_data) + np.array(t3_data), label='t1-t2', color=color_t1)

ax.set_xticks(x)
plt.xticks(fontsize=15, rotation=45)
plt.yticks(fontsize=15)
ax.set_xticklabels(tfs)

ax.legend(fontsize=20)
ax.set_ylabel('Overlapping ratio', fontsize=20)
ax.set_title('scDGRN', fontsize=30)

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
#plt.savefig("./aged_opc/regulon_overlap_CellEvo.pdf",format="pdf")
#plt.savefig("./aged_opc_UMI/regulon_overlap_scDGRN.pdf",format="pdf")
plt.show()
```

### 6. Gene module enrichment analysis
Suppose you currently have a set of regulon modules: g_list. You can directly obtain the enrichment analysis results of this module by gseapy package using the following code.

```python
from matplotlib import rcParams
config = {
    "font.family":'Arial',
    "axes.unicode_minus": False
}
rcParams.update(config)

import pandas as pd
import numpy as np
from gseapy import barplot, dotplot
import gseapy as gp
import matplotlib.pyplot as plt

g_set_specific_raw = []
for tp in range(1,2):
    tf_gene_sets = {}
    df = pd.read_csv('./mouse_brain/cytoscape/aged_opc_UMI/ag_opc_specific_t' + str(tp) + '.csv')

    for _, row in df.iterrows():
        tf = row['TF']
        target = row['Target']
        if tf in tf_gene_sets:
            tf_gene_sets[tf].append(target)
        else:
            tf_gene_sets[tf] = [tf, target]
    g_set_tmp = list(tf_gene_sets.values())
    g_set_specific_raw = g_set_specific_raw + g_set_tmp
    print(f"Total number of gene sets after adding time point {tp}: {len(g_set_specific_raw)}")
g_set_specific = [small_list for small_list in g_set_specific_raw if len(small_list) >= 10]

gene_list_total = g_set_specific

pv_list = []
enrich_term_list = []
module_term_num_list = []
overlap_list = []
count = 0
for g_list in gene_list_total:
    enr = gp.enrichr(gene_list=g_list,
                     gene_sets='GO_Biological_Process_2023',
                     organism='Human',
                     cutoff=0.05)

    significant_results = enr.res2d[enr.res2d['Adjusted P-value'] < 0.05]

    #print("Top 5 enriched terms:")
    #print(significant_results['Term'].head(5))
    #print("Top 5 overlap values:")
    #print(significant_results['Overlap'].head(5))

    et = list(significant_results['Term'].values)
    pv = list(significant_results['Adjusted P-value'].values)
    olap = list(significant_results['Overlap'].values)

    pv_list = pv_list + pv
    enrich_term_list = enrich_term_list + et
    overlap_list = overlap_list + olap

    GO_num = len(enr.res2d[enr.res2d['Adjusted P-value'] < 0.05])
    module_term_num_list.append(GO_num)
    #print(f"Number of significantly enriched terms: {GO_num}")

    if GO_num > 0:
        count = count + 1

significant_rio = count/len(gene_list_total)
print(f"Precision (modules with significant enrichment): {round(significant_rio,2)}")

df = pd.DataFrame({'p-value': pv_list})
df_et = pd.DataFrame({'term': enrich_term_list})
df_term_num = pd.DataFrame({'term_num': module_term_num_list})

overlap_list_new = [round(eval(i),2) for i in overlap_list]
print(f"Processed overlap values (rounded to 2 decimals): {overlap_list_new}")

stage_df = pd.DataFrame({'term':enrich_term_list,'overlap':overlap_list_new,'p-value':pv_list})
print("Enrichment results for all modules:")
print(stage_df)

df_result = pd.DataFrame({'Term': enrich_term_list,'Overlap':overlap_list})
print("Enrichment terms and their overlap values:")
print(df_result)
```

### 8. Validation of dynamic GRNs via TF regulatory activity and motif accessibility analysis

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['font.sans-serif'] = ['Arial']
plt.rcParams['axes.unicode_minus'] = False

COLOR_TARGET = '#1f77b4'
COLOR_MOTIF = '#d62728'

def load_all_grn_data(grn_file_paths):
    print("Preloading GRN network files (once only)...")
    grn_data = {}
    for tp, path in grn_file_paths.items():
        try:
            df = pd.read_csv(path)
            grn_data[tp] = df
            print(f"  - {tp} loaded successfully: {len(df)} edges")
        except Exception as e:
            print(f"  - {tp} load failed: {e}")
            grn_data[tp] = pd.DataFrame(columns=['TF', 'Target'])
    return grn_data

def get_tf_metrics(target_tf, time_points, grn_data, motif_path_template):
    summary_results = []
    tp_to_suffix = {'T1': 't0', 'T2': 't1', 'T3': 't2', 'T4': 't3'}

    for tp in time_points:
        df_grn = grn_data.get(tp)
        if not df_grn.empty:
            target_count = df_grn[df_grn['TF'] == target_tf]['Target'].nunique()
        else:
            target_count = 0

        suffix = tp_to_suffix.get(tp, 't0')
        motif_file = motif_path_template.format(tf=target_tf, suffix=suffix)

        try:
            df_motif = pd.read_csv(motif_file, index_col=0)
            avg_motif = df_motif.iloc[:, 0].mean()
        except Exception:
            avg_motif = 0

        summary_results.append({
            "TimePoint": tp,
            "Target_Count": target_count,
            "Motif_Activity": avg_motif
        })

    return pd.DataFrame(summary_results)

def plot_9_tfs_dynamics(tf_list, grn_data, motif_path_template):
    time_points = ['T1', 'T2', 'T3', 'T4']
    fig, axes = plt.subplots(3, 3, figsize=(20, 12), sharex=True)
    axes = axes.flatten()

    print("\nStarting to plot panels...")

    for idx, tf in enumerate(tf_list):
        ax1 = axes[idx]
        df = get_tf_metrics(tf, time_points, grn_data, motif_path_template)

        ln1 = ax1.plot(df['TimePoint'], df['Target_Count'], color=COLOR_TARGET, marker='o',
                       linewidth=2.5, markersize=8, label='Targets (GRN)')

        ax1.set_title(tf, fontsize=16, fontweight='bold', pad=10)
        ax1.tick_params(axis='y', labelcolor=COLOR_TARGET, labelsize=10)

        if idx % 3 == 0:
            ax1.set_ylabel('TF Regulatory Activity', color=COLOR_TARGET, fontsize=14, fontweight='bold')

        ax2 = ax1.twinx()
        ln2 = ax2.plot(df['TimePoint'], df['Motif_Activity'], color=COLOR_MOTIF, marker='s',
                       linestyle='--', linewidth=2.5, markersize=8, label='Motif Activity')

        ax2.tick_params(axis='y', labelcolor=COLOR_MOTIF, labelsize=10)
        ax2.grid(False)

        if idx % 3 == 2:
            ax2.set_ylabel('Motif Accessibility', color=COLOR_MOTIF, fontsize=14, fontweight='bold')

        lns = ln1 + ln2
        labs = [l.get_label() for l in lns]

        ax1.grid(True, axis='x', linestyle=':', alpha=0.5)

    plt.suptitle("The dynamic changes in the TF regulatory activity and motif accessibility", fontsize=25, y=0.98, fontweight='bold')
    plt.tight_layout(rect=[0, 0, 1, 0.96])
    plt.show()

target_tfs = ['HSF2', 'CUX1', 'SCRT2', 'DLX6', 'GLIS2', 'KLF8', 'NFIB', 'HEY2', 'DLX1']

grn_files = {
    'T1': './mouse_brain/cytoscape/aged_opc_UMI/ag_opc_prediction1_3000_nolog2.csv',
    'T2': './mouse_brain/cytoscape/aged_opc_UMI/ag_opc_prediction2_3000_nolog2.csv',
    'T3': './mouse_brain/cytoscape/aged_opc_UMI/ag_opc_prediction3_3000_nolog2.csv',
    'T4': './mouse_brain/cytoscape/aged_opc_UMI/ag_opc_prediction4_3000_nolog2.csv'
}

motif_path_template = './mouse_brain/scATAC/data/{tf}_motif_activity_{suffix}.csv'

loaded_grn_data = load_all_grn_data(grn_files)
plot_9_tfs_dynamics(target_tfs, loaded_grn_data, motif_path_template)
```

### 9. Assessment of consistency between edge dynamic regulatory strength and target gene accessibility

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import pearsonr, zscore
from collections import Counter

plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['font.sans-serif'] = ['Arial']
plt.rcParams['axes.unicode_minus'] = False

def load_and_normalize_grn(file_path):
    try:
        df = pd.read_csv(file_path)
        cols = df.columns
        weight_col = None
        for c in cols:
            if c.lower() in ['importance', 'weight', 'score', 'coef']:
                weight_col = c
                break
        if weight_col is None:
            weight_col = cols[2]

        df = df.rename(columns={cols[0]: 'TF', cols[1]: 'Target', weight_col: 'Score'})
        df['Edge_ID'] = df['TF'] + "->" + df['Target']

        df['Rank'] = df['Score'].rank(ascending=False)
        max_rank = df['Rank'].max()
        if max_rank > 1:
            df['Norm_Edge_Strength'] = 1 - (df['Rank'] - 1) / (max_rank - 1)
        else:
            df['Norm_Edge_Strength'] = 1.0

        return df[['Edge_ID', 'TF', 'Target', 'Norm_Edge_Strength']]
    except Exception as e:
        print(f"Error loading GRN {file_path}: {e}")
        return pd.DataFrame()

def load_atac_activity(file_path):
    try:
        df = pd.read_csv(file_path, index_col=0)
        gene_means = df.mean(axis=1)
        return gene_means
    except Exception as e:
        print(f"Error loading ATAC {file_path}: {e}")
        return pd.Series(dtype=float)

time_points = ['T1', 'T2', 'T3', 'T4']

grn_files = {
    'T1': './mouse_brain/cytoscape/aged_opc_UMI/ag_opc_prediction1_3000_nolog2.csv',
    'T2': './mouse_brain/cytoscape/aged_opc_UMI/ag_opc_prediction2_3000_nolog2.csv',
    'T3': './mouse_brain/cytoscape/aged_opc_UMI/ag_opc_prediction3_3000_nolog2.csv',
    'T4': './mouse_brain/cytoscape/aged_opc_UMI/ag_opc_prediction4_3000_nolog2.csv'
}

atac_files = {
    'T1': './mouse_brain/scATAC/aged_activity0.csv',
    'T2': './mouse_brain/scATAC/aged_activity1.csv',
    'T3': './mouse_brain/scATAC/aged_activity2.csv',
    'T4': './mouse_brain/scATAC/aged_activity3.csv'
}

print("1. Loading GRN Data...")
grn_data = {}
all_edges_list = []

for tp in time_points:
    df = load_and_normalize_grn(grn_files[tp])
    if not df.empty:
        grn_data[tp] = df.set_index('Edge_ID')
        all_edges_list.extend(df['Edge_ID'].tolist())
    else:
        grn_data[tp] = pd.DataFrame()

edge_counts = Counter(all_edges_list)
selected_edges = {edge for edge, count in edge_counts.items() if count >= 4}

print(f"   Found {len(selected_edges)} edges present in all 4 time points.")

print("   Generating Randomized Control GRN...")
grn_data_random = {}
np.random.seed(42)

for tp in time_points:
    if not grn_data[tp].empty:
        df_rand = grn_data[tp].copy()
        shuffled_scores = df_rand['Norm_Edge_Strength'].sample(frac=1, random_state=42).values
        df_rand['Norm_Edge_Strength'] = shuffled_scores
        grn_data_random[tp] = df_rand
    else:
        grn_data_random[tp] = pd.DataFrame()

print("2. Loading ATAC Gene Activity Data...")
atac_data = {}
for tp in time_points:
    series = load_atac_activity(atac_files[tp])
    atac_data[tp] = series

print("3. Building Analysis Matrix...")

analysis_results = []
edge_info_map = {}
for tp in time_points:
    if tp in grn_data and not grn_data[tp].empty:
        temp_df = grn_data[tp].reset_index()
        current_map = dict(zip(temp_df['Edge_ID'], zip(temp_df['TF'], temp_df['Target'])))
        edge_info_map.update(current_map)

for edge in selected_edges:
    if edge not in edge_info_map: continue
    tf_name, target_name = edge_info_map[edge]

    if all(target_name in atac_data[tp].index for tp in time_points):
        edge_strengths = []
        rand_strengths = []
        target_activities = []

        for tp in time_points:
            if edge in grn_data[tp].index:
                strength = grn_data[tp].loc[edge, 'Norm_Edge_Strength']
            else:
                strength = 0.0

            if edge in grn_data_random[tp].index:
                r_strength = grn_data_random[tp].loc[edge, 'Norm_Edge_Strength']
            else:
                r_strength = 0.0

            edge_strengths.append(strength)
            rand_strengths.append(r_strength)
            target_activities.append(atac_data[tp].loc[target_name])

        if np.std(edge_strengths) > 1e-9 and np.std(target_activities) > 1e-9:
            corr, p = pearsonr(edge_strengths, target_activities)

            if np.std(rand_strengths) > 1e-9:
                rand_corr, rand_p = pearsonr(rand_strengths, target_activities)
            else:
                rand_corr, rand_p = 0, 1.0

            analysis_results.append({
                'Edge': edge,
                'TF': tf_name,
                'Target': target_name,
                'Correlation': corr,
                'P_value': p,
                'Strength_Trend': edge_strengths,
                'Rand_Strength_Trend': rand_strengths,
                'Activity_Trend': target_activities
            })

df_results = pd.DataFrame(analysis_results)
print(f"   Analyzed {len(df_results)} valid edges.")

if not df_results.empty:
    df_pos = df_results[df_results['Correlation'] > 0]
    df_neg = df_results[df_results['Correlation'] < 0]

    def flatten_data_for_plot(df_subset, strength_col='Strength_Trend'):
        all_strengths_z = []
        all_activities_z = []

        for idx, row in df_subset.iterrows():
            s_trend = np.array(row[strength_col])
            a_trend = np.array(row['Activity_Trend'])

            if np.std(s_trend) > 1e-9 and np.std(a_trend) > 1e-9:
                all_strengths_z.extend(zscore(s_trend))
                all_activities_z.extend(zscore(a_trend))

        return pd.DataFrame({
            'Edge_Strength_Z': all_strengths_z,
            'ATAC_Activity_Z': all_activities_z
        })

    flat_pos_real = flatten_data_for_plot(df_pos, 'Strength_Trend')
    flat_pos_rand = flatten_data_for_plot(df_pos, 'Rand_Strength_Trend')

    flat_neg_real = flatten_data_for_plot(df_neg, 'Strength_Trend')
    flat_neg_rand = flatten_data_for_plot(df_neg, 'Rand_Strength_Trend')

    fig, axes = plt.subplots(1, 2, figsize=(14, 6), sharey=True)

    def get_p_text(p_val):
        return "P < 1e-300" if p_val == 0 else f"P = {p_val:.2e}"

    if not flat_pos_real.empty:
        ax = axes[0]

        sns.regplot(data=flat_pos_rand, x='Edge_Strength_Z', y='ATAC_Activity_Z', ax=ax,
                    scatter_kws={'alpha': 0.05, 's': 10, 'color': 'gray'},
                    line_kws={'color': 'gray', 'linewidth': 2, 'linestyle': '--', 'label': 'Random Shuffled'})

        r_rand, p_rand = pearsonr(flat_pos_rand['Edge_Strength_Z'], flat_pos_rand['ATAC_Activity_Z'])

        sns.regplot(data=flat_pos_real, x='Edge_Strength_Z', y='ATAC_Activity_Z', ax=ax,
                    scatter_kws={'alpha': 0.1, 's': 15, 'color': '#d62728'},
                    line_kws={'color': '#8c0900', 'linewidth': 2, 'label': 'Observed GRN'})

        r_real, p_real = pearsonr(flat_pos_real['Edge_Strength_Z'], flat_pos_real['ATAC_Activity_Z'])

        stats_text = (
            f"Observed:\n  r = {r_real:.3f}\n  {get_p_text(p_real)}\n\n"
            f"Random:\n  r = {r_rand:.3f}\n  {get_p_text(p_rand)}"
        )
        ax.text(0.05, 0.95, stats_text, transform=ax.transAxes, fontsize=11, verticalalignment='top',
                bbox=dict(facecolor='white', alpha=0.9, edgecolor='black', boxstyle='round,pad=0.5'))

        ax.set_title(f'Putative Activators (n={len(df_pos)})', fontsize=14, fontweight='bold')
        ax.set_xlabel('Edge Strength (Z-score)', fontsize=12)
        ax.set_ylabel('Target Gene Accessibility (Z-score)', fontsize=12)
        ax.legend(loc='lower right')

    if not flat_neg_real.empty:
        ax = axes[1]

        sns.regplot(data=flat_neg_rand, x='Edge_Strength_Z', y='ATAC_Activity_Z', ax=ax,
                    scatter_kws={'alpha': 0.05, 's': 10, 'color': 'gray'},
                    line_kws={'color': 'gray', 'linewidth': 2, 'linestyle': '--', 'label': 'Random Shuffled'})

        r_rand, p_rand = pearsonr(flat_neg_rand['Edge_Strength_Z'], flat_neg_rand['ATAC_Activity_Z'])

        sns.regplot(data=flat_neg_real, x='Edge_Strength_Z', y='ATAC_Activity_Z', ax=ax,
                    scatter_kws={'alpha': 0.1, 's': 15, 'color': '#1f77b4'},
                    line_kws={'color': '#003366', 'linewidth': 2, 'label': 'Observed GRN'})

        r_real, p_real = pearsonr(flat_neg_real['Edge_Strength_Z'], flat_neg_real['ATAC_Activity_Z'])

        stats_text = (
            f"Observed:\n  r = {r_real:.3f}\n  {get_p_text(p_real)}\n\n"
            f"Random:\n  r = {r_rand:.3f}\n  {get_p_text(p_rand)}"
        )
        ax.text(0.05, 0.95, stats_text, transform=ax.transAxes, fontsize=11, verticalalignment='top',
                bbox=dict(facecolor='white', alpha=0.9, edgecolor='black', boxstyle='round,pad=0.5'))

        ax.set_title(f'Putative Repressors (n={len(df_neg)})', fontsize=14, fontweight='bold')
        ax.set_xlabel('Edge Strength (Z-score)', fontsize=12)
        ax.set_ylabel('')
        ax.legend(loc='lower left')

    plt.tight_layout()
    plt.show()

else:
    print("No valid edges found for analysis.")
```

### 10. Consistency analysis of time-specific regulatory edges and target gene accessibility


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import zscore, ttest_ind

plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['font.sans-serif'] = ['Arial']
plt.rcParams['axes.unicode_minus'] = False

def load_and_normalize_grn(file_path):
    try:
        df = pd.read_csv(file_path)
        cols = df.columns
        weight_col = None
        for c in cols:
            if c.lower() in ['importance', 'weight', 'score', 'coef']:
                weight_col = c
                break
        if weight_col is None:
            weight_col = cols[2]

        df = df.rename(columns={cols[0]: 'TF', cols[1]: 'Target', weight_col: 'Score'})
        df['Edge_ID'] = df['TF'] + "->" + df['Target']

        df['Rank'] = df['Score'].rank(ascending=False)
        max_rank = df['Rank'].max()
        if max_rank > 1:
            df['Norm_Edge_Strength'] = 1 - (df['Rank'] - 1) / (max_rank - 1)
        else:
            df['Norm_Edge_Strength'] = 1.0

        return df[['Edge_ID', 'TF', 'Target', 'Norm_Edge_Strength']]
    except Exception as e:
        print(f"Error loading GRN {file_path}: {e}")
        return pd.DataFrame()

def load_atac_activity(file_path):
    try:
        df = pd.read_csv(file_path, index_col=0)
        gene_means = df.mean(axis=1)
        return gene_means
    except Exception as e:
        print(f"Error loading ATAC {file_path}: {e}")
        return pd.Series(dtype=float)

time_points = ['T1', 'T2', 'T3', 'T4']

grn_files = {
    'T1': './mouse_brain/cytoscape/aged_opc_UMI/ag_opc_prediction1_3000_nolog2.csv',
    'T2': './mouse_brain/cytoscape/aged_opc_UMI/ag_opc_prediction2_3000_nolog2.csv',
    'T3': './mouse_brain/cytoscape/aged_opc_UMI/ag_opc_prediction3_3000_nolog2.csv',
    'T4': './mouse_brain/cytoscape/aged_opc_UMI/ag_opc_prediction4_3000_nolog2.csv'
}

atac_files = {
    'T1': './mouse_brain/scATAC/aged_activity0.csv',
    'T2': './mouse_brain/scATAC/aged_activity1.csv',
    'T3': './mouse_brain/scATAC/aged_activity2.csv',
    'T4': './mouse_brain/scATAC/aged_activity3.csv'
}

print("1. Loading GRN Data...")
grn_data = {}
edge_occurrence = {}

for tp in time_points:
    df = load_and_normalize_grn(grn_files[tp])
    if not df.empty:
        grn_data[tp] = df.set_index('Edge_ID')
        for edge in df['Edge_ID']:
            if edge not in edge_occurrence:
                edge_occurrence[edge] = []
            edge_occurrence[edge].append(tp)
    else:
        grn_data[tp] = pd.DataFrame()

unique_edges_map = {}
for edge, tps in edge_occurrence.items():
    if len(tps) == 1:
        unique_edges_map[edge] = tps[0]

print(f"   Found {len(unique_edges_map)} edges that appear in EXACTLY ONE time point.")

print("2. Loading ATAC Gene Activity Data...")
atac_data = {}
for tp in time_points:
    series = load_atac_activity(atac_files[tp])
    atac_data[tp] = series

print("3. Analyzing Unique Edges vs Target Activity...")

analysis_results = []
valid_edge_count = 0

edge_info_map = {}
for tp in time_points:
    if not grn_data[tp].empty:
        temp_df = grn_data[tp].reset_index()
        current_map = dict(zip(temp_df['Edge_ID'], zip(temp_df['TF'], temp_df['Target'])))
        edge_info_map.update(current_map)

for edge, specific_tp in unique_edges_map.items():
    if edge not in edge_info_map: continue

    tf_name, target_name = edge_info_map[edge]

    if all(target_name in atac_data[tp].index for tp in time_points):
        raw_activities = []
        for tp in time_points:
            raw_activities.append(atac_data[tp].loc[target_name])

        if np.std(raw_activities) == 0:
            z_scores = [0, 0, 0, 0]
        else:
            z_scores = zscore(raw_activities)

        edge_strength = grn_data[specific_tp].loc[edge, 'Norm_Edge_Strength']

        tp_index = time_points.index(specific_tp)
        activity_at_specific_tp = z_scores[tp_index]

        other_indices = [i for i in range(4) if i != tp_index]
        activity_at_other_tps = [z_scores[i] for i in other_indices]
        avg_activity_others = np.mean(activity_at_other_tps)

        analysis_results.append({
            'Edge': edge,
            'Specific_TP': specific_tp,
            'Target': target_name,
            'Edge_Strength': edge_strength,
            'Activity_Z_Specific': activity_at_specific_tp,
            'Activity_Z_Others_Avg': avg_activity_others,
            'Activity_Trend_Z': z_scores
        })
        valid_edge_count += 1

df_results = pd.DataFrame(analysis_results)
print(f"   Analyzed {len(df_results)} valid unique edges.")

if not df_results.empty:
    plot_data_specific = pd.DataFrame({
        'Z_Score': df_results['Activity_Z_Specific'],
        'Condition': 'Edge Present (Specific TP)'
    })

    plot_data_others = pd.DataFrame({
        'Z_Score': df_results['Activity_Z_Others_Avg'],
        'Condition': 'Edge Absent (Other TPs)'
    })

    plot_data = pd.concat([plot_data_specific, plot_data_others])

    plt.figure(figsize=(6, 5))
    sns.violinplot(x='Condition', y='Z_Score', data=plot_data, palette=['#529ed5', '#95a5a6'], inner='quartile')

    t_stat, p_val = ttest_ind(df_results['Activity_Z_Specific'], df_results['Activity_Z_Others_Avg'])

    plt.title(f'Edge Present vs. Absent\n(Time-specific Edges Only, n={len(df_results)})', fontsize=14)
    plt.ylabel('Target Gene Accessibility (Z-score)')
    plt.xlabel('')

    y_max = plot_data['Z_Score'].max()
    plt.text(0.5, y_max, f'P-value = {p_val:.2e}', ha='center', fontsize=12, fontweight='bold')

    plt.tight_layout()
    plt.show()

    df_heatmap_source = df_results.copy()
    df_heatmap_source['Specificity_Score'] = (
            df_heatmap_source['Activity_Z_Specific'] - df_heatmap_source['Activity_Z_Others_Avg']
    )
    df_top_list = []
    for tp in time_points:
        sub = df_heatmap_source[df_heatmap_source['Specific_TP'] == tp] \
            .sort_values('Specificity_Score', ascending=False).head(15)
        df_top_list.append(sub)
    df_heatmap_source = pd.concat(df_top_list)

    heatmap_matrix = []
    heatmap_labels = []
    row_colors = []
    tp_colors = {'T1': '#1f77b4', 'T2': '#ff7f0e', 'T3': '#2ca02c', 'T4': '#d62728'}

    for _, row in df_heatmap_source.iterrows():
        heatmap_matrix.append(row['Activity_Trend_Z'])
        tf, tgt = edge_info_map[row['Edge']]
        tp = row['Specific_TP']
        heatmap_labels.append(f"{tp} | {tf}->{tgt}")
        row_colors.append(tp_colors[tp])

    heatmap_df = pd.DataFrame(heatmap_matrix, columns=time_points, index=heatmap_labels)

    plt.figure(figsize=(6, 10))
    ax = sns.heatmap(
        heatmap_df,
        cmap="RdBu_r",
        center=0,
        linewidths=0,
        linecolor='white',
        cbar_kws={'label': 'Z-score', 'shrink': 0.6},
        xticklabels=True,
        yticklabels=True
    )

    ax.set_title('Target Gene Accessibility of Time-specific Regulatory Edges ', pad=20, fontweight='bold')
    ax.set_xlabel('Time Points')
    ax.set_ylabel('')
    plt.yticks(rotation=0, fontsize=7)
    plt.xticks(rotation=0)

    from matplotlib.patches import Patch
    legend_elements = [Patch(facecolor=tp_colors[tp], label=tp) for tp in time_points]

    plt.tight_layout()
    plt.show()

    print("\n=== Summary Statistics ===")
    print(f"Mean Z-score when Edge is Present: {df_results['Activity_Z_Specific'].mean():.3f}")
    print(f"Mean Z-score when Edge is Absent:  {df_results['Activity_Z_Others_Avg'].mean():.3f}")
    print("Interpretation: Positive Z-score means activity is above the gene's average across time.")

else:
    print("No valid unique edges found where target gene exists in ATAC data.")
```
